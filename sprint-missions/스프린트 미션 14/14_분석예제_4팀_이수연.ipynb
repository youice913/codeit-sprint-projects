### 문제8: 이상치 처리하기

- 데이터에 이상치가 있는지 확인해보세요. 확인 후, 전처리가 필요하다면 진행해주세요.

```python
# 수치형 변수별로 boxplot을 그려 이상치 시각화
plt.rc('font', family='Malgun Gothic')
plt.figure(figsize=(20, 12))
sns.boxplot(data=data.select_dtypes(include=[np.number]), orient='h', palette="Set2")
plt.title('수치형 변수별 이상치 시각화 (Boxplot)', fontsize=18, fontweight='bold', color='#333366')
plt.xlabel('값', fontsize=14)
plt.ylabel('변수', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=data[['PURCHASES', 'ONEOFF_PURCHASES', 'INSTALLMENTS_PURCHASES']],
    orient='h',
    palette="Set2"
)
plt.title('구매 관련 변수의 이상치 시각화 (Boxplot)', fontsize=16, fontweight='bold', color='#333366')
plt.xlabel('값', fontsize=12)
plt.ylabel('변수', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# 이상치 제거 (전처리 진행)
# IQR 기반 이상치 제거 
def remove_outliers_iqr(df, columns):
    df_clean = df.copy()
    for col in columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]
    return df_clean

# PURCHASES, ONEOFF_PURCHASES, INSTALLMENTS_PURCHASES 변수의 이상치 제거
outlier_cols = ['PURCHASES', 'ONEOFF_PURCHASES', 'INSTALLMENTS_PURCHASES']
data = remove_outliers_iqr(data, outlier_cols)
```

```python
# 이상치 제거 후 시각화
plt.figure(figsize=(20, 12))
sns.set(style="whitegrid", palette="muted")
box = sns.boxplot(
    data=data[outlier_cols],
    orient='h',
    palette="Spectral"
)
plt.title('이상치 제거 후 구매 관련 변수의 이상치 시각화 (Boxplot)', fontsize=18, fontweight='bold', color='#333366')
plt.xlabel('값', fontsize=14)
plt.ylabel('변수', fontsize=14)
plt.tight_layout()
plt.show()
```

## 클러스터 분석 (11문제)

###문제10: K-means 클러스터링 모델 학습
- K-means 알고리즘을 사용하여 클러스터링 모델을 학습하세요. 클러스터의 수는 3로 설정하세요.

```python
from sklearn.cluster import KMeans

# K-means 클러스터링 (클러스터 수 3)
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
```

```python
# 클러스터 할당
df_kmeans = X.copy()

df_kmeans['Cluster_Kmeans'] = kmeans.predict(X)
df_kmeans['Cluster_Kmeans'].value_counts()
```

### 문제11 : 클러스터링 결과 시각화 (2D)
- 두 개의 주요 변수(BALANCE, PURCHASES)를 사용하여 클러스터링 결과를 2D로 시각화하세요.

```python
plt.rc('font', family='Malgun Gothic')	

plt.figure(figsize=(10, 7))
sns.scatterplot(
    x=X['BALANCE'],
    y=X['PURCHASES'],
    hue=df_kmeans['Cluster_Kmeans'],
    palette='Set1',
    alpha=0.7
)
plt.title('K-means 클러스터링 결과 (BALANCE vs PURCHASES)', fontsize=16, fontweight='bold')
plt.xlabel('BALANCE (표준화)')
plt.ylabel('PURCHASES (표준화)')
plt.legend(title='Cluster')
plt.tight_layout()
plt.show()
```

### 문제12: 클러스터 수 결정 (Elbow Method)
- Elbow Method를 사용하여 최적의 클러스터 수를 결정하세요.

```python
# Elbow Method를 사용하여 최적의 클러스터 수 찾기
inertia = []
cluster_range = range(1, 11)
for k in cluster_range:
    model = KMeans(n_clusters=k, random_state=42)
    model.fit(X)
    inertia.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(cluster_range, inertia, marker='o')
plt.title('Elbow Method for Optimal k', fontsize=16, fontweight='bold')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.xticks(cluster_range)
plt.grid(True)
plt.show()
```

- 4 정도가 최적으로 보입니다. k=4 이후로는 점차 완만한 감소만 이어집니다.

```python
from sklearn.metrics import silhouette_score, calinski_harabasz_score, davies_bouldin_score
from scipy.cluster.hierarchy import cut_tree
import pandas as pd

# KMeans (k=4)
kmeans_4 = KMeans(n_clusters=4, random_state=42)
kmeans_4_labels = kmeans_4.fit_predict(X)

# GMM (k=4)
gmm_4 = GaussianMixture(n_components=4, random_state=42)
gmm_4_labels = gmm_4.fit_predict(X)

# 계층적 클러스터링 (k=4)
hierarchy_4_labels = cut_tree(Z, n_clusters=4).reshape(-1)

# DBSCAN (클러스터 개수 자동, 노이즈 제외)
dbscan_labels = dbscan_tuned.labels_

def clustering_scores(X, labels, name):
    # DBSCAN의 경우, 클러스터가 2개 미만이면 score 계산 불가
    n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
    if n_clusters < 2:
        return {f"{name}_silhouette": None, f"{name}_calinski": None, f"{name}_davies": None}
    mask = labels != -1 if -1 in labels else slice(None)
    return {
        f"{name}_silhouette": silhouette_score(X[mask], labels[mask]),
        f"{name}_calinski": calinski_harabasz_score(X[mask], labels[mask]),
        f"{name}_davies": davies_bouldin_score(X[mask], labels[mask])
    }

scores = {}
scores.update(clustering_scores(X.values, kmeans_4_labels, "KMeans"))
scores.update(clustering_scores(X.values, gmm_4_labels, "GMM"))
scores.update(clustering_scores(X.values, hierarchy_4_labels, "Hierarchy"))
scores.update(clustering_scores(X.values, dbscan_labels, "DBSCAN"))

pd.DataFrame([scores])
```

####  각 방법의 특징 및 장단점

- **K-means**
  - **특징:** 데이터가 균등하게 분포된 경우 효과적, 해석이 직관적
  - **장점:** 속도 빠름, 대규모 데이터에 적합, 결과가 안정적
  - **단점:** 클러스터 수(k) 사전 지정 필요, 구형(원형) 클러스터에만 강함, 이상치에 민감

- **계층적 클러스터링**
  - **특징:** 덴드로그램으로 계층적 구조 시각화 가능
  - **장점:** 클러스터 수를 나중에 결정 가능, 소규모 데이터에 적합
  - **단점:** 대규모 데이터에 비효율적, 노이즈에 약함

- **DBSCAN**
  - **특징:** 밀도 기반, 노이즈(이상치) 자동 탐지
  - **장점:** 클러스터 수 지정 불필요, 임의 모양의 클러스터 탐지, 노이즈 분리
  - **단점:** 파라미터(eps, min_samples) 튜닝 필요, 밀도 차이 큰 데이터에 약함, 소수 클러스터가 잘 안 잡힘

- **GMM (Gaussian Mixture Model)**
  - **특징:** 클러스터가 타원형/확률적 분포일 때 강점
  - **장점:** 소속 확률 제공, 복잡한 분포에 유연
  - **단점:** 초기값에 민감, 클러스터 수 지정 필요, 계산량 많음

#### 🏁 결론 및 추천
- **K-means**는 전체적으로 균형 잡힌 클러스터 분포와 높은 평가 지표를 보임(실루엣, 칼린스키 등).
- **DBSCAN**은 노이즈 탐지에 강점이 있으나, 주요 클러스터 외 소수 클러스터가 적고, 평가 지표가 일부 낮음(특히 Calinski).
- **계층적 클러스터링**은 시각화와 해석에 유리하나, 대규모 데이터에는 비효율적.
- **GMM**은 복잡한 분포에 유연하지만, 실루엣 점수 등에서 K-means보다 낮음.

> **실무 적용:**  
> - 데이터가 비교적 균등하고, 클러스터 수를 명확히 정할 수 있다면 **K-means** 추천  
> - 이상치/노이즈가 많거나, 클러스터 모양이 불규칙하다면 **DBSCAN**  
> - 계층적 관계나 시각화가 중요하다면 **계층적 클러스터링**  
> - 데이터가 여러 분포(타원형 등)로 섞여 있다면 **GMM**

### 종합 해석 및 실무적 시사점

- **PCA 2D 데이터**는 K-means, GMM, DBSCAN 모두에서 Silhouette, Calinski, Davies-Bouldin 등 주요 평가 지표가 원본 데이터보다 더 우수하게 나타남. 즉, 차원 축소를 통해 데이터의 주요 구조가 더 명확하게 드러나고, 클러스터링 품질이 향상될 수 있음을 시사함.
- **K-means**: PCA 2D에서 가장 높은 Silhouette/Calinski, 가장 낮은 Davies-Bouldin을 기록, 군집 품질이 가장 뛰어남.
- **DBSCAN**: 노이즈 탐지에 강점, PCA 2D에서 노이즈가 크게 줄고 군집 품질도 향상됨.
- **GMM**: PCA 2D에서 성능이 크게 개선됨.
- **계층적 클러스터링**: 차원 축소 전후로 큰 차이는 없으나, 해석과 시각화에 유리.

#### 실무 적용 전략
- **정밀한 타깃팅**이 필요하다면 원본 데이터 기반 클러스터링, **시각화/직관적 해석**이 중요하다면 PCA 등 차원 축소 후 클러스터링 추천.
- **여러 방법을 병행**해 결과를 비교·보완하는 것이 가장 효과적.
- **PCA 2D**는 해석과 시각화, 군집 품질 모두에서 강점이 있으므로, 실무에서도 적극 활용할 가치가 높음.
