### **결측치, 이상치 로그변환**

> 참고로, 로그 변환을 적용한 경우와 적용하지 않은 경우 모두 실험을 진행하였으나, 최종 클러스터링 성능(실루엣 계수 기준)은 로그 변환을 적용하지 않은 데이터가 더 우수한 것으로 확인되었습니다.

```python
# 원본 데이터 크기 확인
original_shape = data.shape

# 복사본 생성 및 결측치 처리
cleaned_test = data.copy()
cleaned_test = cleaned_test.dropna(subset=['CREDIT_LIMIT'])
cleaned_test = cleaned_test.fillna(cleaned_test[['MINIMUM_PAYMENTS', 'CREDIT_LIMIT']].median())

# 이상치 제거 함수
def remove_outliers(df, col):
    q3 = df[col].quantile(0.75)
    q1 = df[col].quantile(0.25)
    IQR = q3 - q1
    lower_limit = q1 - 1.5 * IQR
    upper_limit = q3 + 1.5 * IQR
    return df[(df[col] > lower_limit) & (df[col] < upper_limit)]

cols = ['CASH_ADVANCE', 'CASH_ADVANCE_FREQUENCY', 'INSTALLMENTS_PURCHASES', 'MINIMUM_PAYMENTS']
for col in cols:
    cleaned_test = remove_outliers(cleaned_test, col)

# 전처리 전/후 shape 출력
print("Before:", original_shape)
print("After :", cleaned_test.shape)

data_clean = cleaned_test.copy()
```

```python
# 원본 데이터 복제본 생성
data_clean = data.copy()

# CREDIT_LIMIT 결측치가 있는 행 제거
data_clean = data_clean.dropna(subset=['CREDIT_LIMIT'])

# MINIMUM_PAYMENTS 결측치는 중앙값으로 대체
min_pay_median = data_clean['MINIMUM_PAYMENTS'].median()
data_clean['MINIMUM_PAYMENTS'] = data_clean['MINIMUM_PAYMENTS'].fillna(min_pay_median)

# 로그 변환 대상 변수 목록
log_cols = [
    'BALANCE', 'PURCHASES', 'ONEOFF_PURCHASES', 'INSTALLMENTS_PURCHASES',
    'CASH_ADVANCE', 'PAYMENTS', 'MINIMUM_PAYMENTS', 'CREDIT_LIMIT',
    'CASH_ADVANCE_TRX', 'PURCHASES_TRX'
]

# 로그 변환 (log1p: log(1+x))
for col in log_cols:
    data_clean[col] = np.log1p(data_clean[col])

data_clean.head()
```

```python
# 클린 데이터 복사본 생성
corr_data = data_clean.copy()

corr_data = corr_data.drop(columns='CUST_ID')

# 상관계수 행렬 계산 (수치형 변수만)
corr_matrix = corr_data.corr()

# 상관관계 히트맵 시각화
plt.figure(figsize=(14, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', square=True, linewidths=.5)
plt.title('수치형 변수 간 상관관계 히트맵', fontsize=16)
plt.tight_layout()
plt.show()
```

  → 신용한도가 높을수록 잔액, 구매, 결제 금액도 함께 증가하는 경향이 있습니다.  
- `ONEOFF_PURCHASES`와 `INSTALLMENTS_PURCHASES`는 각각 `PURCHASES`와 강한 양의 상관관계를 보입니다.  
  → 총구매액은 두 변수의 합으로 구성되어 있어, 변수 정의가 반영된 결과로 보입니다.  
- `PURCHASES_TRX`, `PURCHASES_FREQUENCY`, `ONEOFF_PURCHASES_FREQUENCY` 등 빈도 관련 변수들끼리도 높은 상관관계를 보입니다.  
- `TENURE`는 대부분의 변수와 상관관계가 거의 없습니다.  
  → 고객 대다수가 동일한 기간(12개월)을 보유하고 있기 때문으로 보입니다.  
- `PRC_FULL_PAYMENT`은 다른 변수들과 뚜렷한 상관관계를 보이지 않습니다.  
- 전반적으로 금액/빈도 관련 변수들은 서로 상관성이 높으며, 일부 변수(`TENURE`, `PRC_FULL_PAYMENT`)는 독립적인 특성을 갖는 것으로 보입니다.  
- 분석 및 모델링 시 다중공선성이 발생할 수 있는 변수 조합에 주의가 필요해 보입니다.

## **최종 전처리**

### 최종 데이터 전처리 요약

- 분석에 앞서, 신용카드 고객 데이터의 품질을 점검한 결과, `CREDIT_LIMIT` 변수에서 1건, `MINIMUM_PAYMENTS` 변수에서 313건의 결측치가 확인되었습니다. 이에 따라 `CREDIT_LIMIT` 결측치는 해당 행을 제거하였고, `MINIMUM_PAYMENTS` 결측치는 중앙값으로 대체하였습니다.

- 변수 분포를 확인한 결과, 대부분의 금액 및 거래 횟수 관련 변수에서 우측 편향이 뚜렷하게 나타나 로그 변환을 적용하였으며, 이후 모든 수치형 변수에 대해 표준화(standardization)를 수행하였습니다.
- 최종적으로, 식별자 역할의 `CUST_ID` 변수는 분석에 불필요하다고 판단되어 제거하였습니다.

이와 같은 일련의 전처리 과정을 통해 데이터의 품질을 개선하고, 클러스터링 등 후속 분석에 적합한 형태로 데이터를 준비하였습니다.

### K-Means

```python
# 1. K-Means: 적정 k 찾기 (Elbow/inertia plot)
inertias = []
K_range = range(2, 11)
for k in K_range:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(scaled)
    inertias.append(km.inertia_)

plt.plot(K_range, inertias, marker='o')
plt.xlabel('k')
plt.ylabel('Inertia')
plt.title('K-Means Elbow Method')
plt.show()

"""적정 k 값은 4로 보임"""
```

```python
# 최적 k로 클러스터링
k = 4  # 예시, elbow plot 보고 결정
kmeans = KMeans(n_clusters=k, random_state=42)
kmeans_labels = kmeans.fit_predict(scaled)
```

```python
# KMeans 클러스터별 개수 집계

kmeans_counts = pd.Series(kmeans_labels).value_counts().sort_index()
kmeans_counts.name = 'Count'
kmeans_counts.index.name = 'Cluster'
kmeans_counts.to_frame()
```

```python
import seaborn as sns

plt.figure(figsize=(10, 7))
sns.scatterplot(
    x=cleaned['BALANCE'],
    y=cleaned['PURCHASES'],
    hue=kmeans_labels,
    palette='Set1',
    alpha=0.7,
    legend='full'
)
plt.title('K-means(4) 클러스터링 결과 (BALANCE vs PURCHASES)', fontsize=16, fontweight='bold')
plt.xlabel('BALANCE (log-transformed)')
plt.ylabel('PURCHASES (log-transformed)')
plt.legend(title='Cluster')
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(scaled, kmeans_labels)
print("실루엣 계수:", score)
```

### 계층적 군집 

```python
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
from collections import Counter
from sklearn.metrics import silhouette_score
```

```python
# 계층적 군집 학습 (ward linkage)
Z = linkage(scaled, method='ward')

# 덴드로그램 시각화
plt.figure(figsize=(12, 5))
dendrogram(Z, truncate_mode='level', p=10)
plt.title('Hierarchical Clustering Dendrogram (truncated)')
plt.xlabel('Sample Index')
plt.ylabel('Distance')
plt.tight_layout()
plt.show()
```

```python
# 클러스터 할당 (k개 군집)
agg_scipy_labels = fcluster(Z, k, criterion='maxclust')

# 각 계층별(클러스터별) 개수 카운트
cluster_counts = Counter(agg_scipy_labels)
print("클러스터별 개수:", dict(sorted(cluster_counts.items())))
```

```python
# 5. 클러스터링 결과 시각화 (PCA 2D)
plt.figure(figsize=(7, 5))
plt.scatter(scaled[:, 0], scaled[:, 1], c=agg_scipy_labels, cmap='tab10', s=10, alpha=0.7)
plt.title('Hierarchical Clustering Result (Ward, k={})'.format(k))
plt.xlabel('Feature 1 (scaled)')
plt.ylabel('Feature 2 (scaled)')
plt.tight_layout()
plt.show()

# 6. 실루엣 스코어 계산
score_agg = silhouette_score(scaled, agg_scipy_labels)
print("실루엣 계수 (scipy 계층적 군집):", score_agg)
```

## 결론
DBSCAN(PCA 2D) 알고리즘을 적용한 결과, 데이터 내에서 서로 다른 특성을 가진 여러 고객 그룹이 식별되었습니다. 각 그룹의 주요 특성과 이에 적합한 마케팅 방안은 다음과 같이 정리됩니다.

- **Group 0 (주요 대다수 그룹, 평균적 사용 고객)**
  - 특징: 신용카드 사용 실적이 평균적이며, 잔액과 구매액 모두 중간 수준으로 나타납니다.
  - 마케팅 방안: 일반적인 리워드 프로그램, 포인트 적립, 생활밀착형 혜택 등 보편적 프로모션을 제공하여 충성도를 유지하고 점진적 이용 증대를 유도하는 것이 효과적일 것으로 보입니다.

- **Group 1 (고액/활발 이용 고객군)**
  - 특징: 잔액, 구매액, 결제액 등 주요 지표가 전체적으로 높은 상위 그룹으로 확인됩니다.
  - 마케팅 방안: 프리미엄 카드 업그레이드, 한도 증액, 맞춤형 VIP 혜택, 고액 결제 리워드 등 차별화된 서비스를 제공하여 이탈을 방지하고 추가 매출을 유도하는 전략이 적합할 것으로 판단됩니다.

- **Group 2 (저이용/비활성 고객군)**
  - 특징: 카드 사용 실적이 매우 낮거나, 거래가 거의 없는 고객들로 구성된 것으로 보입니다.
  - 마케팅 방안: 연회비 면제, 첫 결제 이벤트, 맞춤형 할인 쿠폰 등 이용 활성화를 위한 인센티브를 제공하여 휴면 방지 및 재이용을 유도하는 것이 바람직할 것으로 보입니다.

- **Group 3 (특정 패턴/이상치 고객군)**
  - 특징: 일부 변수에서 극단적인 값(예: 현금서비스 집중, 특정 거래 패턴 등)이 관찰되는 소수 그룹으로 확인됩니다.
  - 마케팅 방안: 리스크 관리 차원의 모니터링 강화, 맞춤형 상담, 필요시 신용관리 서비스 연계 등 개별적 접근이 필요할 것으로 판단됩니다.

이와 같이 DBSCAN(PCA 2D) 기반의 군집별 특성을 반영한 차별화된 마케팅 전략을 수립함으로써, 고객 만족도 제고와 기업의 수익성 향상을 동시에 도모할 수 있을 것으로 기대됩니다.
